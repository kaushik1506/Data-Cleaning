# **Data Cleaning**




```python
df_copy = df.copy()
```

```python
df_copy['Reviews'].unique()
```

```python
# df_copy['Reviews'].astype(int)

# returns a value error
#ValueError: invalid literal for int() with base 10: '3.0M'
```

```python
df_copy['Reviews'].str.isnumeric().sum()
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
df_copy = df_copy.drop(10472,axis=0)
df_copy.shape
```

```python
df_copy['Reviews'] = df_copy['Reviews'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size'] = df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size'] = df_copy['Size'].str.replace('M','000')
df_copy['Size'] = df_copy['Size'].str.replace('k','')
df_copy['Size'] = df_copy['Size'].astype(float)
```

```python
df_copy['Size'].unique()
```

```python
df_copy.info()
```

```python
df_copy.head()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Installs'] = df_copy['Installs'].str.replace('+','')
df_copy['Installs'] = df_copy['Installs'].str.replace(',','')
df_copy['Installs'] = df_copy['Installs'].astype(int)
```

```python
df_copy['Installs']
```

```python
df_copy.info()
```

```python
df_copy.head(3)
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Price'] = df_copy['Price'].str.replace('$','')
df_copy['Price'] = df_copy['Price'].astype(float)
```

```python
df_copy.info()
```

```python
df_copy['Content Rating'].unique()
```

```python
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.info()
```

```python
df_copy = df_copy.drop('Last Updated',axis=1)
```

```python
df_copy.info()
```
